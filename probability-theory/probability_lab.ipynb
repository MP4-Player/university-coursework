1)Исследование двумерного закона распределения

а)Построить таблицу совместного распределения двух случайных величин (вероятность рассматривать статистическую):
X - средняя оценка первой контрольной работы
Y - средняя оценка второй контрольной работы
Z - средняя оценка третьей (или более) контрольной работы [применяется для альтернативной таблицы при наличии соответствующего задания]

б)Построить одномерные законы распределения

с)Построить условные законы распределения


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np


file_path = '51.xlsx'  
data = pd.read_excel(file_path, skiprows=2)  

data.columns = ['ID', 'Пол', 'Вариант', '1_общ', '1_замена', '1_по_час', '1_рац', '1_триг', '1_иррац', '1_доп', 'Результат1',
                '2_вариант', '2_текст', '2_параметр', '2_поляр', '2_вращ', '2_дуга', '2_несоб', '2_доп', 'Результат2',
                '3_с_разд', '3_однород', '3_лин', '3_Бернулли', '3_доп', 'Результат3']

X = data['Результат1']  
Y = data['Результат2']  
Z = data['Результат3']  
ID = data['ID']  
Gender = data['Пол']  
Variant = data['Вариант']  


def build_distributions(df1, df2, bins=5):
    bins_x = np.linspace(0.0, 1.0, bins + 1, endpoint=True)  
    bins_y = np.linspace(0.0, 1.0, bins + 1, endpoint=True)  
    df1_binned = pd.cut(df1['result'], bins=bins_x)
    df2_binned = pd.cut(df2['result'], bins=bins_y)

    merged_df = pd.DataFrame({
        'id_student': df1['id_student'],
        'bin_x': df1_binned,
        'bin_y': df2_binned
    })

    joint_table = merged_df.groupby(['bin_x', 'bin_y']).size().unstack(fill_value=0)

    joint_probabilities = joint_table / len(merged_df)

    distribution_x = df1_binned.value_counts(sort=False).sort_index() / len(df1)
    distribution_y = df2_binned.value_counts(sort=False).sort_index() / len(df2)

    conditional_x = joint_probabilities.divide(distribution_y, axis=1)
    conditional_y = joint_probabilities.divide(distribution_x, axis=0)

    return joint_probabilities, distribution_x, distribution_y, conditional_x, conditional_y

# Преобразуем данные в формат, подходящий для функции build_distributions
df1 = pd.DataFrame({'id_student': ID, 'result': X})
df2 = pd.DataFrame({'id_student': ID, 'result': Y})
df3 = pd.DataFrame({'id_student': ID, 'result': Z})


# Построение распределений для X и Y
joint_probabilities_XY, distribution_X, distribution_Y, conditional_X_given_Y, conditional_Y_given_X = build_distributions(df1, df2)

# Построение распределений для X и Z
joint_probabilities_XZ, _, _, conditional_X_given_Z, conditional_Z_given_X = build_distributions(df1, df3)

# Построение распределений для Y и Z
joint_probabilities_YZ, _, _, conditional_Y_given_Z, conditional_Z_given_Y = build_distributions(df2, df3)

# Построение распределений для   YX
joint_probabilities_YX, distribution_Y, distribution_X, conditional_Y_given_X, conditional_X_given_Y = build_distributions(df2, df1)

# Построение распределений для  ZX
joint_probabilities_ZX, _, _, conditional_Z_given_X , conditional_X_given_Z= build_distributions(df3, df1)

# Построение распределений для ZY
joint_probabilities_ZY, _, _, conditional_Z_given_Y , conditional_Y_given_Z = build_distributions(df3, df2)


print("Совместное распределение X и Y:")
print(joint_probabilities_XY)

print("\nСовместное распределение X и Z:")
print(joint_probabilities_XZ)

print("\nСовместное распределение Y и Z:")
print(joint_probabilities_YZ)

print("\nУсловные распределения X при условии Y:")
print(conditional_X_given_Y)

print("\nУсловные распределения Y при условии X:")
print(conditional_Y_given_X)

print("\nУсловные распределения X при условии Z:")
print(conditional_X_given_Z)

print("\nУсловные распределения Z при условии X:")
print(conditional_Z_given_X)

print("\nУсловные распределения Y при условии Z:")
print(conditional_Y_given_Z)

print("\nУсловные распределения Z при условии Y:")
print(conditional_Z_given_Y)



# Визуализация совместных распределений
plt.figure(figsize=(18, 12))  

# Совместное распределение X и Y
plt.subplot(3, 3, 1)  
sns.heatmap(joint_probabilities_XY.fillna(0), annot=True, cmap='YlGnBu', cbar=True)
plt.title('Совместное распределение X и Y')
plt.xlabel('Y (Результат второй контрольной)')
plt.ylabel('X (Результат первой контрольной)')

# Совместное распределение X и Z
plt.subplot(3, 3, 2)  
sns.heatmap(joint_probabilities_XZ.fillna(0), annot=True, cmap='YlGnBu', cbar=True)
plt.title('Совместное распределение X и Z')
plt.xlabel('Z (Результат третьей контрольной)')
plt.ylabel('X (Результат первой контрольной)')

# Совместное распределение Y и Z
plt.subplot(3, 3, 3)  
sns.heatmap(joint_probabilities_YZ.fillna(0), annot=True, cmap='YlGnBu', cbar=True)
plt.title('Совместное распределение Y и Z')
plt.xlabel('Z (Результат третьей контрольной)')
plt.ylabel('Y (Результат второй контрольной)')

# Совместное распределение Y и X
plt.subplot(3, 3, 4)  
sns.heatmap(joint_probabilities_YX.fillna(0), annot=True, cmap='YlGnBu', cbar=True)
plt.title('Совместное распределение Y и X')
plt.xlabel('X (Результат первой контрольной)')
plt.ylabel('Y (Результат второй контрольной)')

# Совместное распределение Z и X
plt.subplot(3, 3, 5) 
sns.heatmap(joint_probabilities_ZX.fillna(0), annot=True, cmap='YlGnBu', cbar=True)
plt.title('Совместное распределение Z и X')
plt.xlabel('X (Результат первой контрольной)')
plt.ylabel('Z (Результат третьей контрольной)')

# Совместное распределение Z и Y
plt.subplot(3, 3, 6)  
sns.heatmap(joint_probabilities_ZY.fillna(0), annot=True, cmap='YlGnBu', cbar=True)
plt.title('Совместное распределение Z и Y')
plt.xlabel('Y (Результат второй контрольной)')
plt.ylabel('Z (Результат третьей контрольной)')

plt.tight_layout()
plt.show()


# Визуализация условных распределений
plt.figure(figsize=(18, 12))  # Увеличиваем размер фигуры для лучшей читаемости

# Условные распределения X при условии Y
plt.subplot(3, 3, 1)  # 3 строки, 3 столбца, 1-й график
sns.heatmap(conditional_X_given_Y.fillna(0), annot=True, cmap='coolwarm', cbar=True)
plt.title('Условные распределения X при условии Y')
plt.xlabel('Y (Результат второй контрольной)')
plt.ylabel('X (Результат первой контрольной)')

# Условные распределения Y при условии X
plt.subplot(3, 3, 2)  # 3 строки, 3 столбца, 2-й график
sns.heatmap(conditional_Y_given_X.fillna(0), annot=True, cmap='coolwarm', cbar=True)
plt.title('Условные распределения Y при условии X')
plt.xlabel('X (Результат первой контрольной)')
plt.ylabel('Y (Результат второй контрольной)')

# Условные распределения X при условии Z
plt.subplot(3, 3, 3)  # 3 строки, 3 столбца, 3-й график
sns.heatmap(conditional_X_given_Z.fillna(0), annot=True, cmap='coolwarm', cbar=True)
plt.title('Условные распределения X при условии Z')
plt.xlabel('Z (Результат третьей контрольной)')
plt.ylabel('X (Результат первой контрольной)')

# Условные распределения Z при условии X
plt.subplot(3, 3, 4)  # 3 строки, 3 столбца, 4-й график
sns.heatmap(conditional_Z_given_X.fillna(0), annot=True, cmap='coolwarm', cbar=True)
plt.title('Условные распределения Z при условии X')
plt.xlabel('X (Результат первой контрольной)')
plt.ylabel('Z (Результат третьей контрольной)')

# Условные распределения Y при условии Z
plt.subplot(3, 3, 5)  # 3 строки, 3 столбца, 5-й график
sns.heatmap(conditional_Y_given_Z.fillna(0), annot=True, cmap='coolwarm', cbar=True)
plt.title('Условные распределения Y при условии Z')
plt.xlabel('Z (Результат третьей контрольной)')
plt.ylabel('Y (Результат второй контрольной)')

# Условные распределения Z при условии Y
plt.subplot(3, 3, 6)  # 3 строки, 3 столбца, 6-й график
sns.heatmap(conditional_Z_given_Y.fillna(0), annot=True, cmap='coolwarm', cbar=True)
plt.title('Условные распределения Z при условии Y')
plt.xlabel('Y (Результат второй контрольной)')
plt.ylabel('Z (Результат третьей контрольной)')

plt.tight_layout()
plt.show()

print("DDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDDD")
# a) Построение таблицы совместного распределения X и Y
joint_distribution_XY = pd.crosstab(X, Y, margins=True, margins_name='Всего')
print("Таблица совместного распределения X и Y:")
print(joint_distribution_XY)

# Визуализация совместного распределения X и Y
plt.figure(figsize=(10, 6))
sns.heatmap(joint_distribution_XY, annot=True, fmt='d', cmap='YlGnBu', cbar=True)
plt.title('Совместное распределение X и Y')
plt.xlabel('Y (Результат второй контрольной)')
plt.ylabel('X (Результат первой контрольной)')
plt.show()

# a) Построение таблицы совместного распределения z и Y
joint_distribution_ZY = pd.crosstab(Z, Y, margins=True, margins_name='Всего')
print("Таблица совместного распределения Z и Y:")
print(joint_distribution_ZY)

# Визуализация совместного распределения Z и Y
plt.figure(figsize=(10, 6))
sns.heatmap(joint_distribution_ZY, annot=True, fmt='d', cmap='YlGnBu', cbar=True)
plt.title('Совместное распределение Z и Y')
plt.xlabel('Y (Результат второй контрольной)')
plt.ylabel('Z (Результат третей контрольной)')
plt.show()

# Если есть третья контрольная, построим таблицу для X и Z
if 'Результат3' in data.columns:
    joint_distribution_XZ = pd.crosstab(X, Z, margins=True, margins_name='Всего')
    print("\nТаблица совместного распределения X и Z:")
    print(joint_distribution_XZ)

    # Визуализация совместного распределения X и Z
    plt.figure(figsize=(10, 6))
    sns.heatmap(joint_distribution_XZ, annot=True, fmt='d', cmap='YlGnBu', cbar=True)
    plt.title('Совместное распределение X и Z')
    plt.xlabel('Z (Результат третьей контрольной)')
    plt.ylabel('X (Результат первой контрольной)')
    plt.show()

# b) Построение одномерных законов распределения
# Для X
distribution_X = X.value_counts().sort_index()
print("\nОдномерное распределение X:")
print(distribution_X)

# Визуализация распределения X
plt.figure(figsize=(8, 5))
sns.barplot(x=distribution_X.index, y=distribution_X.values, palette='viridis')
plt.title('Распределение X (Результат первой контрольной)')
plt.xlabel('Оценка')
plt.ylabel('Частота')
plt.show()

# Для Y
distribution_Y = Y.value_counts().sort_index()
print("\nОдномерное распределение Y:")
print(distribution_Y)

# Визуализация распределения Y
plt.figure(figsize=(8, 5))
sns.barplot(x=distribution_Y.index, y=distribution_Y.values, palette='viridis')
plt.title('Распределение Y (Результат второй контрольной)')
plt.xlabel('Оценка')
plt.ylabel('Частота')
plt.show()

# Для Z (если есть)
if 'Результат3' in data.columns:
    distribution_Z = Z.value_counts().sort_index()
    print("\nОдномерное распределение Z:")
    print(distribution_Z)

    # Визуализация распределения Z
    plt.figure(figsize=(8, 5))
    sns.barplot(x=distribution_Z.index, y=distribution_Z.values, palette='viridis')
    plt.title('Распределение Z (Результат третьей контрольной)')
    plt.xlabel('Оценка')
    plt.ylabel('Частота')
    plt.show()

# c) Построение условных законов распределения
# Условные распределения Y при условии z
conditional_distribution_Y_given_Z = pd.crosstab(Z, Y, normalize='index')
print("\nУсловные распределения Y при условии Z:")
print(conditional_distribution_Y_given_Z)

# Визуализация условных распределений Y при условии Z
plt.figure(figsize=(10, 6))
sns.heatmap(conditional_distribution_Y_given_Z, annot=True, cmap='coolwarm', cbar=True)
plt.title('Условные распределения Y при условии Z')
plt.xlabel('Y (Результат второй контрольной)')
plt.ylabel('Z (Результат первой контрольной)')
plt.show()

# Условные распределения X при условии Y
conditional_distribution_X_given_Y = pd.crosstab(X, Y, normalize='columns')
print("\nУсловные распределения X при условии Y:")
print(conditional_distribution_X_given_Y)

# Визуализация условных распределений X при условии Y
plt.figure(figsize=(10, 6))
sns.heatmap(conditional_distribution_X_given_Y, annot=True, cmap='coolwarm', cbar=True)
plt.title('Условные распределения X при условии Y')
plt.xlabel('Y (Результат второй контрольной)')
plt.ylabel('X (Результат первой контрольной)')
plt.show()

# Если есть третья контрольная, построим условные распределения Z при условии X

conditional_distribution_Z_given_X = pd.crosstab(X, Z, normalize='index')
print("\nУсловные распределения Z при условии X:")
print(conditional_distribution_Z_given_X)

# Визуализация условных распределений Z при условии X
plt.figure(figsize=(10, 6))
sns.heatmap(conditional_distribution_Z_given_X, annot=True, cmap='coolwarm', cbar=True)
plt.title('Условные распределения Z при условии X')
plt.xlabel('Z (Результат третьей контрольной)')
plt.ylabel('X (Результат первой контрольной)')
plt.show()

#наоборот
# Условные распределения Y при условии z
conditional_distribution_Z_given_Y = pd.crosstab(Y, Z, normalize='index')
print("\nУсловные распределения Y при условии Z:")
print(conditional_distribution_Z_given_Y)

# Визуализация условных распределений Y при условии Z
plt.figure(figsize=(10, 6))
sns.heatmap(conditional_distribution_Z_given_Y, annot=True, cmap='coolwarm', cbar=True)
plt.title('Условные распределения Y при условии Z')
plt.xlabel('Y (Результат второй контрольной)')
plt.ylabel('Z (Результат первой контрольной)')
plt.show()

# Условные распределения X при условии Y
conditional_distribution_Y_given_X = pd.crosstab(Y, X, normalize='columns')
print("\nУсловные распределения X при условии Y:")
print(conditional_distribution_Y_given_X)

# Визуализация условных распределений X при условии Y
plt.figure(figsize=(10, 6))
sns.heatmap(conditional_distribution_Y_given_X, annot=True, cmap='coolwarm', cbar=True)
plt.title('Условные распределения X при условии Y')
plt.xlabel('Y (Результат второй контрольной)')
plt.ylabel('X (Результат первой контрольной)')
plt.show()

# Если есть третья контрольная, построим условные распределения Z при условии X

conditional_distribution_X_given_Z = pd.crosstab(Z, X, normalize='index')
print("\nУсловные распределения Z при условии X:")
print(conditional_distribution_X_given_Z)

# Визуализация условных распределений Z при условии X
plt.figure(figsize=(10, 6))
sns.heatmap(conditional_distribution_X_given_Z, annot=True, cmap='coolwarm', cbar=True)
plt.title('Условные распределения Z при условии X')
plt.xlabel('Z (Результат третьей контрольной)')
plt.ylabel('X (Результат первой контрольной)')
plt.show()

# d) Анализ по полу
# Распределение X по пол
distribution_X_by_gender = data.groupby('Пол')['Результат1'].value_counts().unstack().fillna(0)
print("\nРаспределение X по пол:")
print(distribution_X_by_gender)

# Визуализация распределения X по пол
plt.figure(figsize=(8, 5))
distribution_X_by_gender.plot(kind='bar', stacked=True, colormap='viridis')
plt.title('Распределение X по полу')
plt.xlabel('Пол')
plt.ylabel('Частота')
plt.legend(title='Оценка')
plt.show()

# Распределение Y по пол
distribution_Y_by_gender = data.groupby('Пол')['Результат2'].value_counts().unstack().fillna(0)
print("\nРаспределение Y по пол:")
print(distribution_Y_by_gender)

# Визуализация распределения Y по пол
plt.figure(figsize=(8, 5))
distribution_Y_by_gender.plot(kind='bar', stacked=True, colormap='viridis')
plt.title('Распределение Y по пол')
plt.xlabel('Пол')
plt.ylabel('Частота')
plt.legend(title='Оценка')
plt.show()


# Распределение z по пол
distribution_Z_by_gender = data.groupby('Пол')['Результат3'].value_counts().unstack().fillna(0)
print("\nРаспределение z по пол:")
print(distribution_Z_by_gender)

# Визуализация распределения z по пол
plt.figure(figsize=(8, 5))
distribution_Z_by_gender.plot(kind='bar', stacked=True, colormap='viridis')
plt.title('Распределение Z по пол')
plt.xlabel('Пол')
plt.ylabel('Частота')
plt.legend(title='Оценка')
plt.show()

# e) Анализ по варианту

# Фильтруем данные: оставляем только тех студентов, у которых есть вариант
data_with_variant = data[data['Вариант'].notna() & (data['Вариант'] != 0)]

# Извлечение столбцов с результатами контрольных работ
X = data_with_variant['Результат1']  # Средняя оценка первой контрольной
Y = data_with_variant['Результат2']  # Средняя оценка второй контрольной
Z = data_with_variant['Результат3']  # Средняя оценка третьей контрольной

# Извлечение столбцов с ID, полом и вариантом
ID = data_with_variant['ID']  # ID студента
Gender = data_with_variant['Пол']  # Пол
Variant = data_with_variant['Вариант']  # Вариант
# Распределение X по варианту
distribution_X_by_variant = data_with_variant.groupby('Вариант')['Результат1'].value_counts().unstack().fillna(0)
print("\nРаспределение X по варианту:")
print(distribution_X_by_variant)

# Визуализация распределения X по варианту
plt.figure(figsize=(8, 5))
distribution_X_by_variant.plot(kind='bar', stacked=True, colormap='viridis')
plt.title('Распределение X по варианту')
plt.xlabel('Вариант')
plt.ylabel('Частота')
plt.legend(title='Оценка')
plt.show()

# Распределение Y по варианту
distribution_Y_by_variant = data_with_variant.groupby('Вариант')['Результат2'].value_counts().unstack().fillna(0)
print("\nРаспределение Y по варианту:")
print(distribution_Y_by_variant)

# Визуализация распределения Y по варианту
plt.figure(figsize=(8, 5))
distribution_Y_by_variant.plot(kind='bar', stacked=True, colormap='viridis')
plt.title('Распределение Y по варианту')
plt.xlabel('Вариант')
plt.ylabel('Частота')
plt.legend(title='Оценка')
plt.show()

# Распределение Z по варианту
#distribution_Z_by_variant = data.groupby('Вариант')['Результат3'].value_counts().unstack().fillna(0)
#print("\nРаспределение Z по варианту:")
#print(distribution_Z_by_variant)

# Визуализация распределения Z по варианту
#plt.figure(figsize=(8, 5))
#distribution_Z_by_variant.plot(kind='bar', stacked=True, colormap='viridis')
#plt.title('Распределение Z по варианту')
#plt.xlabel('Вариант')
#plt.ylabel('Частота')
#plt.legend(title='Оценка')
#plt.show()

# f) Совместный анализ: таблица совместного распределения X и Y с учётом пола
joint_distribution_XY_by_gender = pd.crosstab([X, Gender], Y, margins=True, margins_name='Всего')
print("\nТаблица совместного распределения X и Y с учётом пола:")
print(joint_distribution_XY_by_gender)

# Визуализация совместного распределения X и Y с учётом пола
plt.figure(figsize=(10, 6))
sns.heatmap(joint_distribution_XY_by_gender, annot=True, fmt='d', cmap='YlGnBu', cbar=True)
plt.title('Совместное распределение X и Y с учётом пола')
plt.xlabel('Y (Результат второй контрольной)')
plt.ylabel('X (Результат первой контрольной) и Пол')
plt.show()

#x z

# f) Совместный анализ: таблица совместного распределения X и z с учётом пола
joint_distribution_XZ_by_gender = pd.crosstab([X, Gender], Z, margins=True, margins_name='Всего')
print("\nТаблица совместного распределения X и Z с учётом пола:")
print(joint_distribution_XZ_by_gender)

# Визуализация совместного распределения X и Z с учётом пола
plt.figure(figsize=(10, 6))
sns.heatmap( joint_distribution_XZ_by_gender, annot=True, fmt='.2f', cmap='YlGnBu', cbar=True)
plt.title('Совместное распределение X и Z с учётом пола')
plt.xlabel('z (Результат третей контрольной)')
plt.ylabel('X (Результат первой контрольной) и Пол')
plt.show()

#yz

# f) Совместный анализ: таблица совместного распределения y и z с учётом пола
joint_distribution_YZ_by_gender = pd.crosstab([Y, Gender], Z, margins=True, margins_name='Всего')
print("\nТаблица совместного распределения Y и Z с учётом пола:")
print(joint_distribution_YZ_by_gender)

# Визуализация совместного распределения Y и Z с учётом пола
plt.figure(figsize=(10, 6))
sns.heatmap( joint_distribution_YZ_by_gender, annot=True, fmt='.2f', cmap='YlGnBu', cbar=True)
plt.title('Совместное распределение Y и Z с учётом пола')
plt.xlabel('z (Результат третей контрольной)')
plt.ylabel('Y (Результат второй контрольной) и Пол')
plt.show()
```

2)Вычисление статистических характеристик

а)Найти математическое ожидание для одномерных законов

б)Найти дисперсию для одномерных законов

с)Найти среднеквадратическое отклонение для одномерных законов


```python
import numpy as np

# Функция для вычисления статистических характеристик
def calculate_statistics(values):
    # Математическое ожидание (среднее значение)
    mean = np.mean(values)
    # Дисперсия
    variance = np.var(values, ddof=1)  # Используем ddof=1 для несмещённой дисперсии
    # Среднеквадратическое отклонение
    std_dev = np.std(values, ddof=1)  # Используем ddof=1 для несмещённого СКО
    return mean, variance, std_dev

# Вычисление статистических характеристик для X
mean_X, variance_X, std_dev_X = calculate_statistics(X)
print("\nСтатистические характеристики для X:")
print(f"Математическое ожидание (X): {mean_X:.2f}")
print(f"Дисперсия (X): {variance_X:.2f}")
print(f"Среднеквадратическое отклонение (X): {std_dev_X:.2f}")

# Вычисление статистических характеристик для Y
mean_Y, variance_Y, std_dev_Y = calculate_statistics(Y)
print("\nСтатистические характеристики для Y:")
print(f"Математическое ожидание (Y): {mean_Y:.2f}")
print(f"Дисперсия (Y): {variance_Y:.2f}")
print(f"Среднеквадратическое отклонение (Y): {std_dev_Y:.2f}")

# Вычисление статистических характеристик для Z (если есть)
if 'Результат3' in data.columns:
    mean_Z, variance_Z, std_dev_Z = calculate_statistics(Z)
    print("\nСтатистические характеристики для Z:")
    print(f"Математическое ожидание (Z): {mean_Z:.2f}")
    print(f"Дисперсия (Z): {variance_Z:.2f}")
    print(f"Среднеквадратическое отклонение (Z): {std_dev_Z:.2f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn.linear_model import LinearRegression
from scipy.stats import linregress

# Загрузка данных из Excel-файла, пропускаем первые две строки с заголовками
file_path = '5.xlsx'  # Укажи путь к файлу
data = pd.read_excel(file_path, skiprows=2)  # Пропускаем первые две строки

# Переименовываем столбцы вручную
data.columns = ['ID', 'Пол', 'Вариант', '1_общ', '1_замена', '1_по_час', '1_рац', '1_триг', '1_иррац', '1_доп', 'Результат1',
                '2_вариант', '2_текст', '2_параметр', '2_поляр', '2_вращ', '2_дуга', '2_несоб', '2_доп', 'Результат2',
                '3_с_разд', '3_однород', '3_лин', '3_Бернулли', '3_доп', 'Результат3']

# Извлечение столбцов с результатами контрольных работ
X = data['Результат1'].values.reshape(-1, 1)  # Средняя оценка первой контрольной
Y = data['Результат2'].values.reshape(-1, 1)  # Средняя оценка второй контрольной
Z = data['Результат3'].values  # Средняя оценка третьей контрольной

# Удаляем строки, где Z отсутствует (если третья контрольная не заполнена)
data_filtered = data.dropna(subset=['Результат3'])
X_filtered = data_filtered['Результат1'].values.reshape(-1, 1)
Y_filtered = data_filtered['Результат2'].values.reshape(-1, 1)
Z_filtered = data_filtered['Результат3'].values

# Объединяем X и Y в одну матрицу признаков
XY = np.hstack([X_filtered, Y_filtered])

# Создаем модель линейной регрессии для Z = aX + bY + c
model_z = LinearRegression()
model_z.fit(XY, Z_filtered)

# Получаем коэффициенты регрессии для Z
a, b = model_z.coef_  # Коэффициенты для X и Y
c = model_z.intercept_  # Свободный член

# Уравнение плоскости регрессии: Z = a*X + b*Y + c
print(f"Уравнение плоскости регрессии: Z = {a:.2f}*X + {b:.2f}*Y + {c:.2f}")

# Линейная регрессия между Y и X
slope_yx, intercept_yx, _, _, _ = linregress(X_filtered.flatten(), Y_filtered.flatten())

# Линейная регрессия между Z и X
slope_zx, intercept_zx, _, _, _ = linregress(X_filtered.flatten(), Z_filtered.flatten())

# Линейная регрессия между Y и Z
slope_yz, intercept_yz, _, _, _ = linregress(Y_filtered.flatten(), Z_filtered.flatten())

# Уравнения линий регрессии
print(f"Уравнение линии регрессии между Y и X: Y = {slope_yx:.2f}*X + {intercept_yx:.2f}")
print(f"Уравнение линии регрессии между Z и X: Z = {slope_zx:.2f}*X + {intercept_zx:.2f}")
print(f"Уравнение линии регрессии между Y и Z: Z = {slope_yz:.2f}*Y + {intercept_yz:.2f}")

# Построение 3D графика с точками
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Точки (X, Y, Z)
ax.scatter(X_filtered, Y_filtered, Z_filtered, c='blue', marker='o', label='Точки (X, Y, Z)')

# Построение плоскости регрессии
x_plane = np.linspace(0, 1, 100)
y_plane = np.linspace(0, 1, 100)
x_plane, y_plane = np.meshgrid(x_plane, y_plane)
z_plane = a * x_plane + b * y_plane + c

# Отображение плоскости
ax.plot_surface(x_plane, y_plane, z_plane, alpha=0.5, color='red', label='Плоскость регрессии (XYZ)')

# Построение линии регрессии между Y и X
x_line = np.linspace(0, 1, 100)
y_line = slope_yx * x_line + intercept_yx
z_line = np.full_like(x_line, np.mean(Z_filtered))  # Линия на среднем значении Z
ax.plot(x_line, y_line, z_line, c='green', label='Линия регрессии между Y и X')

# Построение линии регрессии между Z и X
z_line_zx = slope_zx * x_line + intercept_zx
y_line_zx = np.full_like(x_line, np.mean(Y_filtered))  # Линия на среднем значении Y
ax.plot(x_line, y_line_zx, z_line_zx, c='purple', label='Линия регрессии между Z и X')

# Построение линии регрессии между Y и Z
z_line_yz = slope_yz * y_line + intercept_yz
x_line_yz = np.full_like(y_line, np.mean(X_filtered))  # Линия на среднем значении X
ax.plot(x_line_yz, y_line, z_line_yz, c='orange', label='Линия регрессии между Y и Z')

# Настройка графика
ax.set_xlabel('X (Результат первой контрольной)')
ax.set_ylabel('Y (Результат второй контрольной)')
ax.set_zlabel('Z (Результат третьей контрольной)')
ax.set_title('Трёхмерный закон распределения и регрессии')
ax.legend()

# Включение интерактивности для вращения графика
plt.ion()  # Включаем интерактивный режим
plt.show(block=True)  # Указываем, что окно должно оставаться открытым

# Ожидание закрытия окна
plt.ioff()  # Отключаем интерактивный режим после закрытия окна
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Загрузка данных
file_path = '53.xlsx'  
data = pd.read_excel(file_path, skiprows=2)  

# Переименовываем столбцы 
data.columns = ['ID', 'Пол', 'Вариант', '1_общ', '1_замена', '1_по_час', '1_рац', '1_триг', '1_иррац', '1_доп', 'Результат1',
                '2_вариант', '2_текст', '2_параметр', '2_поляр', '2_вращ', '2_дуга', '2_несоб', '2_доп', 'Результат2',
                '3_с_разд', '3_однород', '3_лин', '3_Бернулли', '3_доп', 'Результат3']

# Извлечение столбцов с результатами контрольных работ
X = data['Результат1']  # Средняя оценка первой контрольной
Y = data['Результат2']  # Средняя оценка второй контрольной
Z = data['Результат3']  # Средняя оценка третьей контрольной

# Удаляем строки, где X или Y отсутствуют
data_filtered = data.dropna(subset=['Результат1', 'Результат2', 'Результат3'])
X_filtered = data_filtered['Результат1']
Y_filtered = data_filtered['Результат2']
Z_filtered = data_filtered['Результат3']

# Расчет коэффициента корреляции Пирсона между X и Y
r_XY = np.corrcoef(X_filtered, Y_filtered)[0, 1]

# Расчет коэффициента корреляции Пирсона между X и Z
r_XZ = np.corrcoef(X_filtered, Z_filtered)[0, 1]

# Расчет коэффициента корреляции Пирсона между Y и Z
r_YZ = np.corrcoef(Y_filtered, Z_filtered)[0, 1]

# Вычисление средних и стандартных отклонений
mean_X = np.mean(X_filtered)
mean_Y = np.mean(Y_filtered)
mean_Z = np.mean(Z_filtered)

std_X = np.std(X_filtered, ddof=0)  # ddof=0 для вычисления стандартного отклонения по генеральной совокупности
std_Y = np.std(Y_filtered, ddof=0)
std_Z = np.std(Z_filtered, ddof=0)

# Вычисление ковариации
covariance_XY = np.cov(X_filtered, Y_filtered, ddof=0)[0, 1]  # Берем элемент [0, 1] из матрицы ковариации
covariance_XZ = np.cov(X_filtered, Z_filtered, ddof=0)[0, 1]
covariance_YZ = np.cov(Y_filtered, Z_filtered, ddof=0)[0, 1]

# Создание таблицы с результатами
results = pd.DataFrame({
    'Показатель': ['Среднее X', 'Среднее Y', 'Среднее Z', 'Стандартное отклонение X', 'Стандартное отклонение Y', 'Стандартное отклонение Z', 
                   'Ковариация X-Y', 'Ковариация X-Z', 'Ковариация Y-Z', 
                   'Коэффициент корреляции X-Y', 'Коэффициент корреляции X-Z', 'Коэффициент корреляции Y-Z'],
    'Значение': [mean_X, mean_Y, mean_Z, std_X, std_Y, std_Z, 
                 covariance_XY, covariance_XZ, covariance_YZ, 
                 r_XY, r_XZ, r_YZ]
})

# Вывод таблицы
print(results)

# Вывод результата
print(f"Коэффициент корреляции между X и Y: r = {r_XY:.2f}")
print(f"Коэффициент корреляции между X и Z: r = {r_XZ:.2f}")
print(f"Коэффициент корреляции между Y и Z: r = {r_YZ:.2f}")

# Интерпретация корреляции
def interpret_correlation(r):
    if r > 0.85:
        return "Сильная положительная корреляция"
    elif r < -0.85:
        return "Сильная отрицательная корреляция"
    elif 0.2 < r <= 0.85:
        return "Умеренная положительная корреляция"
    elif -0.85 < r < -0.2:
        return "Умеренная отрицательная корреляция"
    else:
        return "Слабая или отсутствующая корреляция"

print(f"Интерпретация корреляции между X и Y: {interpret_correlation(r_XY)}")
print(f"Интерпретация корреляции между X и Z: {interpret_correlation(r_XZ)}")
print(f"Интерпретация корреляции между Y и Z: {interpret_correlation(r_YZ)}")

# Построение графика рассеяния (scatter plot)
plt.figure(figsize=(12, 6))

# График рассеяния между X и Y
plt.subplot(1, 3, 1)
plt.scatter(X_filtered, Y_filtered, c='blue', marker='o', label='Точки (X, Y)')
plt.title('График рассеяния между X и Y')
plt.xlabel('X (Результат первой контрольной)')
plt.ylabel('Y (Результат второй контрольной)')
plt.grid(True)
plt.legend()

# График рассеяния между X и Z
plt.subplot(1, 3, 2)
plt.scatter(X_filtered, Z_filtered, c='green', marker='o', label='Точки (X, Z)')
plt.title('График рассеяния между X и Z')
plt.xlabel('X (Результат первой контрольной)')
plt.ylabel('Z (Результат третьей контрольной)')
plt.grid(True)
plt.legend()

# График рассеяния между Y и Z
plt.subplot(1, 3, 3)
plt.scatter(Y_filtered, Z_filtered, c='red', marker='o', label='Точки (Y, Z)')
plt.title('График рассеяния между Y и Z')
plt.xlabel('Y (Результат второй контрольной)')
plt.ylabel('Z (Результат третьей контрольной)')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()

# Вычисление матрицы корреляции для всех результатов контрольных работ
correlation_matrix = data_filtered[['Результат1', 'Результат2', 'Результат3']].corr()
print("\nМатрица корреляции для результатов контрольных работ:")
print(correlation_matrix)

# Визуализация матрицы корреляции
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', cbar=True)
plt.title('Матрица корреляции для результатов контрольных работ')
plt.show()
```

4)Построение моделей регрессии  

а)Найти уравнение парной линейной регрессии вида Y = aX + b
X — средняя оценка за первую контрольную
Y - средняя оценка второй контрольной работы
 б)Определить параметры
a - коэффициент наклона
 b - свободный коэффициент
с)Интерпретировать модель: как изменения X влияют на Y


```python

from scipy.stats import linregress


# Построение парной линейной регрессии
slope, intercept, r_value, p_value, std_err = linregress(X_filtered, Y_filtered)

# Вывод результатов
print(f"Уравнение регрессии: Y = {slope:.2f} * X + {intercept:.2f}")
print(f"Коэффициент корреляции (r): {r_value:.2f}")
print(f"p-value: {p_value:.4f}")
print(f"Стандартная ошибка: {std_err:.4f}")

# Интерпретация модели
print(f"\nИнтерпретация:")
print(f"1. Коэффициент наклона (a) = {slope:.2f}: При увеличении оценки за первую контрольную на 1, оценка за вторую контрольную увеличивается на {slope:.2f}.")
print(f"2. Свободный член (b) = {intercept:.2f}: Если оценка за первую контрольную равна 0, то оценка за вторую контрольную составит {intercept:.2f}.")

# Построение графика
plt.figure(figsize=(8, 6))
plt.scatter(X_filtered, Y_filtered, c='blue', label='Данные')
plt.plot(X_filtered, slope * X_filtered + intercept, c='red', label='Линия регрессии')
plt.title('Парная линейная регрессия: Y = aX + b')
plt.xlabel('X (Результат первой контрольной)')
plt.ylabel('Y (Результат второй контрольной)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Загрузка данных из Excel-файла, пропускаем первые две строки с заголовками
file_path = '53.xlsx'  # Укажи путь к файлу
data = pd.read_excel(file_path, skiprows=2)  # Пропускаем первые две строки

# Переименовываем столбцы вручную
data.columns = ['ID', 'Пол', 'Вариант', '1_общ', '1_замена', '1_по_час', '1_рац', '1_триг', '1_иррац', '1_доп', 'Результат1',
                '2_вариант', '2_текст', '2_параметр', '2_поляр', '2_вращ', '2_дуга', '2_несоб', '2_доп', 'Результат2',
                '3_с_разд', '3_однород', '3_лин', '3_Бернулли', '3_доп', 'Результат3']

# Список заданий для каждой контрольной
tasks_control_1 = ['1_общ', '1_замена', '1_по_час', '1_рац', '1_триг', '1_иррац', '1_доп']
tasks_control_2 = ['2_текст', '2_параметр', '2_поляр', '2_вращ', '2_дуга', '2_несоб', '2_доп']
tasks_control_3 = ['3_с_разд', '3_однород', '3_лин', '3_Бернулли', '3_доп']

# Функция для преобразования оценок в 5-балльную систему с ограничением диапазона
def convert_to_5_scale(value):
    # Ограничиваем первичные баллы в диапазоне от 0 до 1
    value = np.clip(value, 0, 1)
    # Переводим в 5-балльную систему
    return value * 5

# Функция для построения гистограммы с интервалами
def plot_histogram(data, column, title, bins=5, include_zeros=True):
    # Фильтрация данных
    if not include_zeros:
        # Исключаем нулевые значения только если отсутствует номер варианта
        data = data[(data[column] > 0) | (data['Вариант'].notna())]
    
    # Преобразование оценок в 5-балльную систему
    data[column] = data[column].apply(convert_to_5_scale)
    
    # Построение гистограммы
    plt.figure(figsize=(8, 6))
    plt.hist(data[column], bins=bins, edgecolor='black', alpha=0.7, range=(0, 5))  # Ограничиваем диапазон от 0 до 5
    plt.title(title)
    plt.xlabel('Оценка (5-балльная шкала)')
    plt.ylabel('Количество студентов')
    plt.xticks(np.arange(0, 5.5, 1))  # Интервалы от 0 до 5 с шагом 1
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

# Построение гистограмм для каждого задания в трех контрольных работах
for control_num, tasks in enumerate([tasks_control_1, tasks_control_2, tasks_control_3], start=1):
    for task_num, column_name in enumerate(tasks, start=1):
        # Проверка наличия столбца
        if column_name in data.columns:
            # Случай 1: Учитываем нулевые значения
            plot_histogram(data, column_name, f'Гистограмма для задания {task_num} в контрольной {control_num} (с нулями)', include_zeros=True)
            
            # Случай 2: Не учитываем нулевые значения
            plot_histogram(data, column_name, f'Гистограмма для задания {task_num} в контрольной {control_num} (без нулей)', include_zeros=False)
        else:
            print(f"Столбец '{column_name}' отсутствует в данных.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Загрузка данных из Excel-файла
file_path = '53.xlsx'  # Укажи путь к файлу
data = pd.read_excel(file_path, skiprows=2)  # Пропускаем первые две строки

# Переименовываем столбцы вручную
data.columns = ['ID', 'Пол', 'Вариант', '1_общ', '1_замена', '1_по_час', '1_рац', '1_триг', '1_иррац', '1_доп', 'Результат1',
                '2_вариант', '2_текст', '2_параметр', '2_поляр', '2_вращ', '2_дуга', '2_несоб', '2_доп', 'Результат2',
                '3_с_разд', '3_однород', '3_лин', '3_Бернулли', '3_доп', 'Результат3']

# Ваши группы заданий
groups = {
    'Интегралы': ['1_общ', '1_замена', '1_по_час', '1_рац', '1_триг', '1_иррац', '1_доп', '2_текст', '2_параметр', '2_поляр', '2_вращ', '2_дуга', '2_несоб', '2_доп'],
    'Тригонометрия': ['1_триг', '2_поляр'],
    'Алгебра': ['1_рац', '1_иррац', '2_несоб'],
    'Геометрия': ['2_поляр', '2_вращ', '2_дуга'],
    'Дифференциальные уравнения': ['3_с_разд', '3_однород', '3_лин', '3_Бернулли'],
    'Дополнительные задания': ['1_доп', '2_доп', '3_доп']
}

# Функция для преобразования оценок в 5-балльную систему с ограничением диапазона
def convert_to_5_scale(value):
    # Ограничиваем первичные баллы в диапазоне от 0 до 1
    value = np.clip(value, 0, 1)
    # Переводим в 5-балльную систему
    return value * 5

# Функция для построения гистограммы для группы заданий
def plot_group_histogram(data, columns, title):
    # Преобразование оценок в 5-балльную систему
    data[columns] = data[columns].applymap(convert_to_5_scale)
    
    # Вычисляем средние значения для каждого задания в группе
    mean_scores = data[columns].mean()
    
    # Построение гистограммы
    plt.figure(figsize=(10, 6))
    plt.bar(mean_scores.index, mean_scores.values, color='skyblue')
    plt.xlabel('Задания')
    plt.ylabel('Средняя оценка (5-балльная шкала)')
    plt.title(f'Средние оценки заданий в группе: {title}')
    plt.xticks(rotation=45)  # Поворачиваем подписи по оси X для удобства чтения
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

# Построение гистограмм для каждой группы
for group_name, columns in groups.items():
    # Проверка наличия столбцов
    if all(col in data.columns for col in columns):
        plot_group_histogram(data, columns, group_name)
    else:
        print(f"Некоторые столбцы для группы '{group_name}' отсутствуют в данных.")
```
