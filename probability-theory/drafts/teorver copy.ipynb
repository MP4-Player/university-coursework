2)Вычисление статистических характеристик

а)Найти математическое ожидание для одномерных законов

б)Найти дисперсию для одномерных законов

с)Найти среднеквадратическое отклонение для одномерных законов


```python
import numpy as np

# Функция для вычисления статистических характеристик
def calculate_statistics(values):
    # Математическое ожидание (среднее значение)
    mean = np.mean(values)
    # Дисперсия
    variance = np.var(values, ddof=1)  # Используем ddof=1 для несмещённой дисперсии
    # Среднеквадратическое отклонение
    std_dev = np.std(values, ddof=1)  # Используем ddof=1 для несмещённого СКО
    return mean, variance, std_dev

# Вычисление статистических характеристик для X
mean_X, variance_X, std_dev_X = calculate_statistics(X)
print("\nСтатистические характеристики для X:")
print(f"Математическое ожидание (X): {mean_X:.2f}")
print(f"Дисперсия (X): {variance_X:.2f}")
print(f"Среднеквадратическое отклонение (X): {std_dev_X:.2f}")

# Вычисление статистических характеристик для Y
mean_Y, variance_Y, std_dev_Y = calculate_statistics(Y)
print("\nСтатистические характеристики для Y:")
print(f"Математическое ожидание (Y): {mean_Y:.2f}")
print(f"Дисперсия (Y): {variance_Y:.2f}")
print(f"Среднеквадратическое отклонение (Y): {std_dev_Y:.2f}")

# Вычисление статистических характеристик для Z (если есть)
if 'Результат3' in data.columns:
    mean_Z, variance_Z, std_dev_Z = calculate_statistics(Z)
    print("\nСтатистические характеристики для Z:")
    print(f"Математическое ожидание (Z): {mean_Z:.2f}")
    print(f"Дисперсия (Z): {variance_Z:.2f}")
    print(f"Среднеквадратическое отклонение (Z): {std_dev_Z:.2f}")
```
