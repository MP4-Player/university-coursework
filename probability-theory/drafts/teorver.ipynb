```python
import pandas as pd

# Загрузка данных из Excel-файла, пропускаем первые две строки с заголовками
file_path = '51.xlsx'  # Укажи путь к файлу
data = pd.read_excel(file_path, skiprows=2)  # Пропускаем первые две строки

# Переименовываем столбцы вручную
data.columns = ['ID', 'Пол', 'Вариант', '1_общ', '1_замена', '1_по_час', '1_рац', '1_триг', '1_иррац', '1_доп', 'Результат1',
                '2_вариант', '2_текст', '2_параметр', '2_поляр', '2_вращ', '2_дуга', '2_несоб', '2_доп', 'Результат2',
                '3_с_разд', '3_однород', '3_лин', '3_Бернулли', '3_доп', 'Результат3']

# Извлечение столбцов с результатами контрольных работ
X = data['Результат1']  # Средняя оценка первой контрольной
Y = data['Результат2']  # Средняя оценка второй контрольной
Z = data['Результат3']  # Средняя оценка третьей контрольной

# Извлечение столбцов с ID, полом и вариантом
ID = data['ID']  # ID студента
Gender = data['Пол']  # Пол
Variant = data['Вариант']  # Вариант

# a) Построение таблицы совместного распределения X и Y
joint_distribution_XY = pd.crosstab(X, Y, margins=True, margins_name='Всего')
print("Таблица совместного распределения X и Y:")
print(joint_distribution_XY)

# Если есть третья контрольная, построим таблицу для X и Z
if 'Результат3' in data.columns:
    joint_distribution_XZ = pd.crosstab(X, Z, margins=True, margins_name='Всего')
    print("\nТаблица совместного распределения X и Z:")
    print(joint_distribution_XZ)

# b) Построение одномерных законов распределения
# Для X
distribution_X = X.value_counts().sort_index()
print("\nОдномерное распределение X:")
print(distribution_X)

# Для Y
distribution_Y = Y.value_counts().sort_index()
print("\nОдномерное распределение Y:")
print(distribution_Y)

# Для Z (если есть)
if 'Результат3' in data.columns:
    distribution_Z = Z.value_counts().sort_index()
    print("\nОдномерное распределение Z:")
    print(distribution_Z)

# c) Построение условных законов распределения
# Условные распределения Y при условии X
conditional_distribution_Y_given_X = pd.crosstab(X, Y, normalize='index')
print("\nУсловные распределения Y при условии X:")
print(conditional_distribution_Y_given_X)

# Условные распределения X при условии Y
conditional_distribution_X_given_Y = pd.crosstab(X, Y, normalize='columns')
print("\nУсловные распределения X при условии Y:")
print(conditional_distribution_X_given_Y)

# Если есть третья контрольная, построим условные распределения Z при условии X
if 'Результат3' in data.columns:
    conditional_distribution_Z_given_X = pd.crosstab(X, Z, normalize='index')
    print("\nУсловные распределения Z при условии X:")
    print(conditional_distribution_Z_given_X)

# d) Анализ по полу
# Распределение X по пол
distribution_X_by_gender = data.groupby('Пол')['Результат1'].value_counts().unstack().fillna(0)
print("\nРаспределение X по пол:")
print(distribution_X_by_gender)

# Распределение Y по пол
distribution_Y_by_gender = data.groupby('Пол')['Результат2'].value_counts().unstack().fillna(0)
print("\nРаспределение Y по пол:")
print(distribution_Y_by_gender)

# e) Анализ по варианту
# Распределение X по варианту
distribution_X_by_variant = data.groupby('Вариант')['Результат1'].value_counts().unstack().fillna(0)
print("\nРаспределение X по варианту:")
print(distribution_X_by_variant)

# Распределение Y по варианту
distribution_Y_by_variant = data.groupby('Вариант')['Результат2'].value_counts().unstack().fillna(0)
print("\nРаспределение Y по варианту:")
print(distribution_Y_by_variant)

# f) Совместный анализ: таблица совместного распределения X и Y с учётом пола
joint_distribution_XY_by_gender = pd.crosstab([X, Gender], Y, margins=True, margins_name='Всего')
print("\nТаблица совместного распределения X и Y с учётом пола:")
print(joint_distribution_XY_by_gender)
```
